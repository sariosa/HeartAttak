```python
import pandas as pd
import numpy as np
df = pd.read_csv('/content/drive/MyDrive/heart.csv')
df.head(20)
```

```python
# columns in an array except column output

X = df.drop('output', axis=1).values
X
```

```python
Y = df['output'].values
Y
```

```python
# Normalize the X data
min_max_scaler = preprocessing.MinMaxScaler()
X_scaled = min_max_scaler.fit_transform(X)
X_scaled

# Normalize the Y data (if needed - Y is categorical in this case)
# Typically, you don't normalize target variables (labels) for classification
# If Y were continuous, you would use the same MinMaxScaler or another normalization method
# Example if Y was continuous:
# Y_scaled = min_max_scaler.fit_transform(Y.reshape(-1, 1))  # Reshape to 2D array
# Y_scaled
```

```python
# plot normalized data in standard deviation
# Calculate the standard deviation of the normalized data
std_dev = np.std(X_scaled, axis=0)

plt.figure(figsize=(10, 6))
plt.bar(range(len(std_dev)), std_dev)
plt.xlabel('Features')
plt.ylabel('Standard Deviation')
plt.title('Standard Deviation of Normalized Features')
plt.xticks(range(len(std_dev)), df.drop('output', axis=1).columns, rotation=45, ha='right')  # Set x-axis labels to feature names
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, Y, test_size=0.2, random_state=42)
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

# **Logistic Regression**

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
logreg = LogisticRegression(max_iter=1000)  # Increased max_iter to ensure convergence
logreg.fit(X_train, y_train)

# Make predictions on the test set
y_pred = logreg.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy}")

# classification report
print(classification_report(y_test, y_pred))

# confusion matrix
conf_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(conf_matrix)


# Example prediction
example_data = X_scaled[0].reshape(1, -1)

# Make the prediction
prediction = logreg.predict(example_data)
print(f"Prediction for example data: {prediction}")


```

```python
# plot confusion metrix

import seaborn as sns


conf_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(conf_matrix)

plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Predicted 0', 'Predicted 1'],
            yticklabels=['Actual 0', 'Actual 1'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```

# **KNN**

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=7)
knn.fit(X_train, y_train)
```

```python
yhat = knn.predict(X_test)
print(yhat[0:5])
print(y_test[0:5])
```

```python
#evaluate accuray

from sklearn.metrics import accuracy_score

knn_accuracy_ytrain = accuracy_score(y_train, knn.predict(X_train))
knn_accuracy = accuracy_score(y_test, yhat)
print(f"KNN Accuracy prediction set: {knn_accuracy}")
print(f"KNN Accuracy train set: {knn_accuracy_ytrain}")
```

```python
# accuracy of knn for other values
k_values = range(1, 11)
accuracies = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    accuracies.append(accuracy)
    print(f"Accuracy for k={k}: {accuracy}")


plt.plot(k_values, accuracies)
plt.xlabel("Number of Neighbors (k)")
plt.ylabel("Accuracy")
plt.title("Accuracy of KNN for Different k Values")
plt.show()
```
